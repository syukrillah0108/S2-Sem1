```python
import pandas as pd

# 1. Memuat Data Awal
data = {
    'ID': ['P01', 'P02', 'P03', 'P04', 'P04'],
    'Umur': [20, 'kosong', 200, 24, 24],
    'Kota': ['Solo', 'solo', 'Bandung', 'Bandung', 'Bandung'],
    'Pendapatan (juta)': [4, 6, 8, 10, 10]
}
df = pd.DataFrame(data)
print("=========================================")
print("1. IDENTIFIKASI MASALAH (DATA AWAL)")
print("=========================================")
print(df)
print("\nMasalah:")
print("- Duplikasi: ID P04 terduplikasi")
print("- Tipe data salah: Umur P02 'kosong'")
print("- Outlier: Umur P03 200")
print("- Inkonsistensi teks: 'Solo' dan 'solo'\n")

# 2. Proses Data Cleaning
# Menghapus baris duplikat
df_cleaned = df.drop_duplicates().reset_index(drop=True)

# Menyeragamkan format teks pada kolom Kota (huruf kapital di awal)
df_cleaned['Kota'] = df_cleaned['Kota'].str.title()

# Menangani nilai tidak valid dan outlier pada kolom Umur
# Mengubah teks 'kosong' menjadi NaN (Not a Number)
df_cleaned['Umur'] = pd.to_numeric(df_cleaned['Umur'], errors='coerce') 
# Mengubah umur 200 (outlier) menjadi NaN
df_cleaned.loc[df_cleaned['Umur'] > 100, 'Umur'] = pd.NA 
# Mengisi nilai NaN dengan rata-rata umur yang valid
df_cleaned['Umur'] = df_cleaned['Umur'].fillna(df_cleaned['Umur'].mean()) 

print("=========================================")
print("2. HASIL DATA CLEANING")
print("=========================================")
print(df_cleaned)
print("\n(Catatan: Duplikat dihapus, 'Solo' diseragamkan, Umur dibersihkan dan diisi rata-rata yaitu 22.0)\n")

# 3. Proses Encoding (One-Hot Encoding)
# Mengubah data nominal 'Kota' menjadi numerik (0 dan 1)
df_encoded = pd.get_dummies(df_cleaned, columns=['Kota'], dtype=int)

print("=========================================")
print("3. HASIL ENCODING (ONE-HOT ENCODING)")
print("=========================================")
print(df_encoded, "\n")

# 4. Perhitungan Min-Max Scaler untuk target pendapatan 6
min_pendapatan = df_encoded['Pendapatan (juta)'].min()
max_pendapatan = df_encoded['Pendapatan (juta)'].max()
target_x = 6

# Rumus X_norm = (x - min) / (max - min)
x_norm = (target_x - min_pendapatan) / (max_pendapatan - min_pendapatan)

print("=========================================")
print("4. PERHITUNGAN MIN-MAX SCALER")
print("=========================================")
print(f"Variabel target (x) = {target_x}")
print(f"Nilai Minimum (min) = {min_pendapatan}")
print(f"Nilai Maksimum (max) = {max_pendapatan}\n")
print("X_norm = (x - min) / (max - min)")
print(f"X_norm = ({target_x} - {min_pendapatan}) / ({max_pendapatan} - {min_pendapatan})")
print(f"X_norm = {target_x - min_pendapatan} / {max_pendapatan - min_pendapatan}")
print(f"X_norm = {x_norm:.4f}")
```
